```python
import pandas as pd
```

```python
#load the data
df = pd.read_csv("patients_results_messy.csv")
```

```python
df.head()
```

```python
# make better column names 
df.columns = ['patient_id', 'date', 'age', 'gender', 'blood_pressure_mm_Hg', 'cholesterol_mg_dl', 'glucose', 'notes']

# if you don't want to write everything out: 
# df.columns = df.columns.str.lower().str.replace(r"[^a-z0-9]+","_", regex=True").str.strip("_")
```

```python
# handling the dates
df['date'] = pd.to_datetime(df['date'], format='mixed')
```

```python
#handling the gender column: make consistent with a regex function
df['gender'] = df['gender'].str.strip().str.lower().map({"male":"m", "female":"f", "m":"m", "f":"f"})

#df['gender'] = df['gender'].str.replace(r'^f.*$', 'female', regex=True)
#df['gender'] = df['gender'].str.replace(r'^m.*$', 'male', regex=True)
```

```python
# make blood pressure a numeric value
df[["systolic","diastolic"]]=df["blood_pressure_mm_Hg"].str.split("/",expand=True)
```

```python
# handling glucose
df["glucose_new"] = pd.to_numeric(df["glucose"], errors='coerce') 
###if character is string it converts to NaN and to numeric it it is a number.
###pandas dataframe assumes 
#flag for missing data: 
df["flag4missing_glucose"]=df["glucose_new"].isna()
```
