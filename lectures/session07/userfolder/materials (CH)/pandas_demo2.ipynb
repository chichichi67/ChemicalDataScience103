```python
import pandas as pd
```

```python
data = pd.read_csv("inventory_solids.csv")
```

```python
data.columns= ["name","sum_formula","amount(g)"]
data
```

```python

data.iloc[:,1]
data.iloc[1,1]= "New Value" #change a value in a certain row and column

data.values[1]# returns all values in certain row

data
```

```python
data["amounts(kg)"]= data["amount(g)"]/1000 #create a new column by performing an operation on an existing column
data
```

```python
#remove stuff
data["amounts(mg)"]= data["amount(g)"]*1000 #create a new column by performing an operation on an existing column

data.dropna #would drop all missing values 

data.drop(["amounts(mg)"],axis=1, inplace = True) #inplace = True means that the changes are made to the original. inplace= false just uses a copy of the data. 
#axis=1 means that we are dropping a column, axis=0 would mean dropping a row
#or
data2=data.drop(["amount(g)"],axis=1, inplace = False) #copy
data2
 
data
```

```python
data2.drop(columns=["amounts(kg)"], inplace = True)
data2
```

```python
data2.loc(len(data))=["sodium chloride", "NaCl", 1000, 1] #add a new row to the data frame. The first argument is the index of the new row, the second argument is a list of values for the new row.
data2.drop(6, inplace = True) #drop a row by index
#or data2.drop(data.index[6], inplace = True) #drop a row by index

data2
```

```python
data3= data2.copy()

data4= pd.concat([data2,data3], ignore_index=False, axis=1)  #concatenate two dataframes. ignore_index=True means that the index will be reset to a continuous range of integers.
data3
data4
```

```python
data4.to_csv("complete inventory.csv", index=False) #save the data frame to a csv file. index=False means that the implicit index will not be saved to the file.
```

```python
data5= pd.concat([data2,data3], ignore_index=False, axis=0)
data5
```

```python
#clean up repetitions etc. 

data5.groupby("name").sum() #group by name and sum the amounts. This will give us the total amount of each chemical in the inventory.
#aslo sums strings, not only numerical values
data5["amounts(kg)"].idxmax()
```

```python
data5.describe() #gives us a summary of the data frame, including count, mean, std, min, max, and quartiles for each numerical column.
```
